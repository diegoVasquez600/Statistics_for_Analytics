```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración visual para que se vea más académico
sns.set_theme(style="whitegrid")

# 1. Cargar y limpiar datos
df = pd.read_csv('incidentes_viales_motos.csv')

def hora_a_decimal(h):
    try:
        partes = str(h).split(':')
        return int(partes[0]) + int(partes[1])/60 + int(partes[2])/3600
    except: return None

df['HORA_DECIMAL'] = df['HORA_INCIDENTE'].apply(hora_a_decimal)
df = df.dropna(subset=['HORA_DECIMAL', 'ZONA'])

# 2. Histograma y Densidad (hora_hist_dens.png)
plt.figure(figsize=(9, 5))
sns.histplot(df['HORA_DECIMAL'], kde=True, color='steelblue', bins=24, stat="density")
plt.title('Histograma y Curva de Densidad - Hora del Incidente', fontsize=14)
plt.xlabel('Hora del día (0-24)', fontsize=12)
plt.ylabel('Densidad', fontsize=12)
plt.tight_layout()
plt.savefig('hora_hist_dens.png', dpi=300)
plt.close()

# 3. Boxplot Continuo (hora_box.png)
plt.figure(figsize=(9, 3))
sns.boxplot(x=df['HORA_DECIMAL'], color='lightsteelblue')
plt.title('Dispersión general de la Hora', fontsize=14)
plt.xlabel('Hora del día (0-24)', fontsize=12)
plt.tight_layout()
plt.savefig('hora_box.png', dpi=300)
plt.close()

# 4. Pie Chart Categórico (zona_pie.png)
plt.figure(figsize=(8, 8))
top_zonas = df['ZONA'].value_counts().head(7)
# Agrupar el resto para que el gráfico no se vea sucio
otras = pd.Series({'OTRAS ZONAS': df['ZONA'].value_counts().iloc[7:].sum()})
datos_pie = pd.concat([top_zonas, otras])

colores = sns.color_palette('pastel')[0:8]
plt.pie(datos_pie, labels=datos_pie.index, autopct='%1.1f%%', startangle=140, colors=colores)
plt.title('Distribución de Incidentes por Comuna (Top 7)', fontsize=14)
plt.tight_layout()
plt.savefig('zona_pie.png', dpi=300)
plt.close()

# 5. Boxplot por Categoría (hora_zona_box.png)
plt.figure(figsize=(11, 6))
zonas_interes = df['ZONA'].value_counts().head(8).index
df_filtrado = df[df['ZONA'].isin(zonas_interes)]

sns.boxplot(data=df_filtrado, x='ZONA', y='HORA_DECIMAL', palette='Set2')
plt.title('Distribución Temporal de Incidentes por Comuna Principal', fontsize=14)
plt.xlabel('Comuna (Zona)', fontsize=12)
plt.ylabel('Hora del día (0-24)', fontsize=12)
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig('hora_zona_box.png', dpi=300)
plt.close()

print("¡Proceso terminado! Sube los 4 archivos .png a Overleaf.")
```
